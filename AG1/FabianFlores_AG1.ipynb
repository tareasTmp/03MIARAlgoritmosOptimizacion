## Torres de Hanoi con Divide y vencerás (MEJORADO) {-}

```python
# Algoritmo mejorado con base en materiales de clase,
# "Khan Academy: Computer science theory"
# y desarrollo creativo  propio.
# Se incluye enlaces en la sección de bibliografía al final
#
# Mejoras:
# - Distribución de la lógica entre funciones complementarias
# - Impresión de cada movimiento para simular los postes en consola

PASO   =  0
POSTES = {}
CLAVES = []

def Definir_Postes( n , desde ):
    global POSTES
    global CLAVES
    POSTES = { 'A':[], 'B':[], 'C':[] }
    POSTES[ desde ] = list( range( n , 0 , -1 ) )
    CLAVES = list( POSTES.keys() )
    print("\n* Posiciones iniciales (número representa tamaño) *")
    Print_postes()

def Print_postes():
    global POSTES
    global PASO
    print(f"Paso: {PASO}")
    for key, value in POSTES.items():
        print(key,value)
    print( "\n" , end="" )
    PASO += 1

def Hallar_pivote( desde , hasta ):
    global POSTES
    global CLAVES
    pivote = next( clave for clave in CLAVES if clave not in ( desde , hasta ))
    return pivote

def Mover_disco( desde , hasta ):
    global POSTES
    POSTES[ hasta ].append( POSTES[ desde ].pop() )
    Print_postes()

def Torres_Hanoi_m( n , desde , hasta ):

    if not POSTES: # Crear postes con discos
        Definir_Postes( n , desde )

    if ( n == 0 ): # Caso base, no se mueve nada
        return

    # Caso General:

    # 1) Encontrar el pivote
    pivote = Hallar_pivote( desde , hasta )

    # 2) Mover (n-1) discos desde 'desde' hasta 'pivote'
    Torres_Hanoi_m( (n-1) , desde , pivote )

    # 3) Mover el disco faltante desde 'desde' hasta 'hasta'
    Mover_disco( desde , hasta )

    #4) Mover los discos presentes en 'pivote' sobre poste 'hasta'
    Torres_Hanoi_m( (n-1) , pivote , hasta )

def Torres_Hanoi_arranque( n , desde , hasta ):
    print("Algoritmo de Torres de Hanoi Mejorado")
    print(f"  Discos(n): {n}   Origen: {desde}   Destino: {hasta}")
    print(f"     Pasos requeridos     2^(n) - 1 = { ( 2**n ) - 1 }")
    global POSTES
    POSTES = {}
    Torres_Hanoi_m( n , desde , hasta )

# Prueba:
Torres_Hanoi_arranque( 5 , "A" , "C" )
```

## Sucesión de Fibonacci {-}

```python
#Sucesión_de_Fibonacci
#https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci
#Calculo del termino n-simo de la suscesión de Fibonacci

# Mejora: en el caso de N<2 se cambia de 'return 1' a 'return N'
# con base en la definición del artículo de Wikipedia, así que
# Fibonnaci(0) = 0 ; Fibonnaci(1) = 1 ; Fibonnaci(2) = 1 ; etc.

def Fibonacci(N:int):
  if N < 2:
    return N
  else:
    return Fibonacci(N-1)+Fibonacci(N-2)

print( "Fibonacci(0) = " , Fibonacci(0) )
print( "Fibonacci(1) = " , Fibonacci(1) )
print( "Fibonacci(2) = " , Fibonacci(2) )
print( "Fibonacci(5) = " , Fibonacci(5) )
print( "Fibonacci(8) = " , Fibonacci(8) )
```

## Devolución de cambio por técnica voraz {-}

```python
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION

cifra = 15
denominacion_monedas = [25,10,5,1]
print(f"En ${cifra} y denominaciones de monedas ${denominacion_monedas}, entregar: #{cambio_monedas( cifra , denominacion_monedas )}")

cifra = 28
denominacion_monedas = [25,10,5,1]
print(f"En ${cifra} y denominaciones de monedas ${denominacion_monedas}, entregar: #{cambio_monedas( cifra , denominacion_monedas )}")
```

## N-Reinas por técnica de vueta atrás {-}

```python
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")

def es_prometedora(SOLUCION,etapa):
  #print(SOLUCION)
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    #print("El valor " + str(SOLUCION[i]) + " está " +  str(SOLUCION.count(SOLUCION[i])) + " veces")
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      #print("Comprobando diagonal de " + str(i) + " y " + str(j))
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]):
        return False
  return True

def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        #escribe(solucion)
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0

print("Soluciones al tablero de reinas de tamaño 4:")
reinas(4)
print("Soluciones al tablero de reinas de tamaño 8:")
reinas(8)
```

## Viaje por el rio. Programación dinámica  {-}

```python
################################################################
def Precios(TARIFAS):
################################################################
  #Total de Nodos
  N = len(TARIFAS[0])

  #Inicialización de la tabla de precios
  PRECIOS = [ [9999]*N for i in [9999]*N] # nxn
  RUTA    = [ [ "" ]*N for i in [ "" ]*N] # nxn

  for i in range(0,N-1):
    #RUTA[i][i]    = i             #Para ir de i a i se "pasa por i"
    #PRECIOS[i][i] = 0          #Para ir de i a i se se paga 0
    for j in range(i+1, N):
      MIN        = TARIFAS[i][j]
      print("La tarifa desde " + str(i) + " hasta " + str(j) + " con un precio de " + str(MIN))
      RUTA[i][j] = i

      for k in range(i, j):
        if PRECIOS[i][k] + TARIFAS[k][j] < MIN:
            MIN = min(MIN, PRECIOS[i][k] + TARIFAS[k][j] )
            RUTA[i][j] = k     #Anota que para ir de i a j hay que pasar por k
            print("Ruta indirecta identificada desde "+str(i)+" hasta "+str(j)+" pasando por "+str(k)+" con un precio de "+str(MIN))

        PRECIOS[i][j] = MIN

  return PRECIOS,RUTA

#Determinar la ruta con Recursividad
def calcular_ruta(RUTA, desde, hasta):
  if desde == hasta:
    #print("Ir a :" + str(desde))
    return ""
  else:
    return str(calcular_ruta( RUTA, desde, RUTA[desde][hasta])) +  \
                '*,*' + \
                str(RUTA[desde][hasta] \
              )
################################################################

##### Definición de variables y Pruebas del algoritmo
TARIFAS = [
            [   0 ,   5 ,   4 ,   3 , 999 , 999 , 999 ],
            [ 999 ,   0 , 999 ,   2 ,   3 , 999 ,  11 ],
            [ 999 , 999 ,   0 ,   1 , 999 ,   4 ,  10 ],
            [ 999 , 999 , 999 ,   0 ,   5 ,   6 ,   9 ],
            [ 999 , 999 , 999 , 999 ,   0 , 999 ,   4 ],
            [ 999 , 999 , 999 , 999 , 999 ,   0 ,   3 ],
            [ 999 , 999 , 999 , 999 , 999 , 999 ,   0 ]
          ]
PRECIOS,RUTA = Precios(TARIFAS)
#print(PRECIOS[0][6])
print("PRECIOS")

for i in range(len(TARIFAS)):
  print(PRECIOS[i])

print("\nRUTA")
for i in range(len(TARIFAS)):
  print(RUTA[i])

print("\nLa ruta es:")
calcular_ruta(RUTA, 0,6)
```
